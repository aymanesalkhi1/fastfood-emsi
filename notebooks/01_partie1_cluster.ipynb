## Driver, executors, MapReduce

Le SparkContext vit dans le driver. Il découpe le job en stages et en tasks. Les executors sont les JVM des deux workers : ils exécutent les tasks et gardent les blocs en cache.

MapReduce réécrit chaque étape sur disque. Le chiffre d'affaires par ville et les itérations de PageRank seraient alors des jobs séparés, avec un fichier entre chaque. Spark ne matérialise que le shuffle et rejoue le lineage si une partition disparaît. MapReduce ne lit pas non plus un topic en continu : le watermark et les fenêtres n'y existent pas.
